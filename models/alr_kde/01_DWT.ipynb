```python
import os

import pandas as pd
import numpy as np
import xarray as xr

import matplotlib.pyplot as plt

from bluemath_tk.datamining.kma import KMA
```

```python
#### Load Data
path_pcs = os.path.join("..", "..", "inputs", "pcs_df.csv")
pcs_df = pd.read_csv(path_pcs, index_col=0, parse_dates=True)
```

#### KMA

```python
kma_ob = KMA(num_clusters=36)
kma_bmus, _ = kma_ob.fit_predict(data=pcs_df, min_number_of_points=50, normalize_data=True)
```

#### Save Classifier

```python
path_kma_ob = os.path.join("..", "..", "outputs", "alr", "classifier.pkl")
kma_ob.save_model(path_kma_ob)
```

```python
pcs_df['kma'] = kma_bmus['kma_bmus']
```

```python
path_pcs_out = os.path.join("..", "..", "outputs", "alr", "pcs_with_kma.csv")
pcs_df.to_csv(path_pcs_out)
```
